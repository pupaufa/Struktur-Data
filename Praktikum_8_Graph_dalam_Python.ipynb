Reperesentasi Graph dengan Adjacency List

```python
# Membuat Graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# menampilkan graf
def print_graph (graf):
  for node in graf:
    print(f"{node} -> {graf[node]}")

print_graph(graph)
```

Traversal Graf menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# contoh pemanggilan
bfs(graph, 'A')
```

Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

# contoh pemanggilan
dfs(graph, 'A')
```

Latihan Personal

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Lengkapi fungsi untuk menghitung jumlah tetangga dari tiap simpul dalam graf
def count_neighbors(graph):
  result = {}
  for node in graph:
    result[node] = len(graph[node])
  return result

print(count_neighbors(graph))
```

Penugasan

```python
from collections import deque

# 1
graph = {
 'V': ['W', 'Z'],
 'W': ['V', 'X', 'Z'],
 'X': ['W', 'Y', 'Z'],
 'Y': ['X', 'Z'],
 'Z': ['W', 'V', 'Y', 'X']
}

def print_graph (graf):
  for node in graf:
    print(f"{node} -> {graf[node]}")

print_graph(graph)

# 2 Implementasi BFS & DFS
# BFS
def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

print('\n=========== BFS =============')
bfs(graph, 'Y')

# DFS
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)


print('\n=========== DFS =============')
dfs(graph, 'Y')

# 3 Modifikasi BFS
def bfs(graph, start):
  visited = set()
  hasil_kunjungan = []
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      hasil_kunjungan.append(vertex)
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

  return hasil_kunjungan

print('\n====== Modifikasi BFS =======')
urutan_bfs = bfs(graph, 'Y')
print("List kunjungan BFS:", urutan_bfs)

# 4
def find_path(graph, start, end, path=[]):
  path = path + [start]
  if start == end:
    return path
  if start not in graph:
      return None
  for neighbor in graph[start]:
      if neighbor not in path:
          new_path = find_path(graph, neighbor, end, path)
          if new_path:
            return new_path
  return None


print('======== FIND PATH ==========')
jalur_start_ke_end = find_path(graph, 'V', 'Y')
print(f"Satu jalur dari V ke Y: {jalur_start_ke_end}")

# 5
def is_connected(graph):
    if not graph:
        return True

    start_node = list(graph.keys())[0]
    visited = set()
    stack = [start_node]

    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    stack.append(neighbor)

    return len(visited) == len(graph)

print('======== IS CONNECTED ========')
print(is_connected(graph))
```
